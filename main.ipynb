```python
import numpy as np


def intensity(t, events, mu, alpha, beta):
    """lambda*(t) = mu + sum_{t_i <= t} alpha * exp(-beta * (t - t_i))"""
    past = np.asarray(events)
    past = past[past <= t]
    if past.size == 0:
        return mu
    return mu + alpha * np.exp(-beta * (t - past)).sum()
```
